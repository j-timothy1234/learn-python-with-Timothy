# SETS

## 1. Introduction to Sets

### A set is a Python collection used to store multiple unique values.

### The important characteristics of a set are:

#### . Sets are unordered.
#### . Sets do not allow duplicate values.
#### . Sets are mutable, meaning we can add or remove elements.
#### . Set elements must be hashable.
#### . Sets are very useful for mathematical operations such as union, intersection, and difference.

#### Example

```python
# A set of numbers

numbers = {10, 20, 30, 40}

print(numbers)
```

## 5. Set Elements Must Be Hashable

### Set elements must be objects that Python can hash.

#### Common valid elements include:

```python
# Set 1
numbers = {1, 2, 3}
print(numbers)
print()

# Set 2
names = {"John", "Sarah"}
print(names)
print()

# Set 3
values = {10.5, 20.5}
print(values)
print()

mixed = {1, "Python", 3.5}
print(mixed)
print()
```

### But mutable collections such as lists cannot normally be elements of a set:
#### Forexample;

```python
# This causes an error

my_set = {[1, 2], [3, 4]}
print(my_set)
```

### Why?

#### Because lists are mutable and therefore not hashable.

#### A tuple can be used:

```python
# A tuple in a set

my_set = {(1, 2), (3, 4)}
print(my_set)
```

## 6. Sets Are Unordered

### A set does not work like a list.

```python
# NUmbers randomly put in a list

numbers = {10, 20, 30, 40}
print(numbers)
```

#### You should not depend on the order in which elements appear.

#### Therefore, this is not appropriate:

```python
numbers[0]
```

#### If you need ordered data and indexing, use a list or tuple.

## 7. Checking Whether an Element Exists

### Use the in operator, returns a boolean value (True or False).

#### Forexample

```python
fruits = {"apple", "banana", "orange"}

print("apple" in fruits)
```

### Another example

```python
print("mango" in fruits)
```

#### This is very useful in real applications.
##### Check example below;

```python
# Suppose Company's users are in a set

registered_users = {"timothy", "melissa", "smart", "natasha"} # set of registered members

username = input("Enter username: ") # get username

if username in registered_users: # check if username is in the set
    print(f"User, {username.title()} already exists in the system")
    
else: # username isn't available
    print(f"{username.title()}, does not exist in the system")
```

## 19. Set Operations

### This is where sets become especially powerful.

#### Suppose we have:

```python
# Set A
A = {1, 2, 3, 4}
print(f"Set A: {A}")
print()

# Set B
B = {3, 4, 5, 6}
print(f"Set B: {B}")
print()
```

#### We can perform mathematical operations on them.

## 20. Union

#### Union combines elements from both sets while removing duplicates. See Example Below;

```python
# UNION of SETs A & B

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

A_U_B = A.union(B)

print(f"Set AUB:  {A_U_B}")
```

#### You can also use |:

```python
A_U_B = A | B
print(f"Set AUB:  {A_U_B}")  # Meaning A U B
```

## 21. Intersection

### Intersection finds elements that are present in both sets.

#### Forexample;

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

A_n_B = A.intersection(B)

print(f"Set AnB:  {A_n_B}")
```

#### You can also use &:
#### See Example Below;

```python
A_n_B = A & B
print(f"Set AnB:  {A_n_B}")
```

### Real-world example

#### Suppose:

```python
# We have Sets for Programmers

python_students = {"timothy", "job", "sandra", "mark", "grace"}

java_students = {"mark", "grace", "peter", "david"}
```

#### Students studying both (intersection):

```python
programmers = python_students.intersection(java_students)
print(programmers)
```

#### OR:

```python
programmers = python_students & java_students
print(programmers)
```

## 22. Difference

### Difference finds elements that exist in one set but not the other.

#### Forexample:

```python
# Set A
A = {1, 2, 3, 4}

# Set B
B = {3, 4, 5, 6}

print(A.difference(B))
```

#### Mind about the arrangement of letters for sets while using difference operation

```python
print(B.difference(A))
```

#### Using the operator (-):

```python
print(A - B)
```

##### Or B-A:

### Important

#### A - B is not necessarily the same as B - A.

```python
print(B - A)
```

## 23. Symmetric Difference

### Symmetric difference gives elements that are in either set, but not both, ie; Intersection Eliminated.

#### Forexample;

```python
# Set A
A = {1, 2, 3, 4}

# Set B
B = {3, 4, 5, 6}

result = A.symmetric_difference(B)

print(result)
```

### Using the Operator (^):

```python
result = A ^ B # Intersection ELiminated
print(result)
```

## 24. Summary of Set Operations

### Given:

```python
# Set A
A = {1, 2, 3, 4}

# Set B
B = {3, 4, 5, 6}
```

| Operation            | Python   |
| -------------------- | -------- |
| Union                | `A \| B` |
| Intersection         | `A & B`  |
| Difference           | `A - B`  |
| Reverse Difference   | `B - A`  |
| Symmetric Difference | `A ^ B`  |


## 25. Subsets

### Set A is a subset of Set B if every element in A also exists in B.

#### Forexample

```python
# Set A
A = {1, 2}

# Set B
B = {1, 2, 3, 4}
```

### Check if Subset;

```python
print(A.issubset(B))
```

### You can also use: 

```python
print(A <= B)
```

## 26. Supersets

### A set is a superset if it contains all elements of another set.

#### Forexample:

```python
# Set A
A = {1, 2}

# Set B
B = {1, 2, 3, 4}

print(B.issuperset(A))
```

### You can also use:

```python
print(B >= A)
```

### What if not elements are their, See example below;

```python
# Set A
A = {1, 2, 7}

# Set B
B = {1, 2, 3, 4}

print(B.issuperset(A))
```

#### Returns a 'False', since 7 is not in Set B

## 27. Disjoint Sets

### Two sets are disjoint when they have no common elements.

#### Forexample:

```python
# Set A
A = {1, 2, 3}

# Set B
B = {4, 5, 6}

print(A.isdisjoint(B))
```

### What if they have an element in common, See example below.

```python
# Set A
A = {1, 2, 3, 4}

# Set B
B = {4, 5, 6}

print(A.isdisjoint(B))
print(A )
```

#### Returns a 'False' since they have a '4' in common.

## 31. Set Comprehension

### Just like list comprehensions, Python allows set comprehensions.

#### Example:

```python
numbers = {1, 2, 3, 4, 5}

squares = {number ** 2 for number in numbers}

print(squares)
```

### OR: Iterate through the set

```python
numbers = {1, 2, 3, 4, 5}
print(f"Set of Numbers: {numbers}")
print()

print("Their Respective Squares.")

for number in numbers:
    squares = number ** 2
    print(squares, end = " ")
    
```

### Set Comprehension With a Condition
#### Forexample

```python
numbers = {1, 2, 3, 4, 5, 6, 8, 10, 12}

even_numbers = {
    number
    for number in numbers
    if number % 2 == 0
}

print(even_numbers)
print(len(even_numbers))
```

### OR: Iterate through the set

```python
numbers = {1, 2, 3, 4, 5, 6}
print(f"Set Of Numbers: {numbers}")
print()

print("Even Numbers in It:")

for even in numbers:
    if even % 2 == 0:
        print(even, end = " ")
```
