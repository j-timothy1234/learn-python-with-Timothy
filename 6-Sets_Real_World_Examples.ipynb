## 37. Real-World Example: Student Courses

#### Suppose students are registered for different courses.

````python
python_students = {
    "John",
    "Sarah",
    "Mark",
    "Grace"
}
print(f"Python Students: {python_students}")

print()

java_students = {
    "Mark",
    "Grace",
    "Peter",
    "David"
}
print(f"Java Students: {java_students}")
print()
````

#### Students taking both:

````python
both_courses = python_students & java_students
print(both_courses)
````

#### Students taking python only:

````python
python_only = python_students - java_students
````

#### Students taking atleast one:

````python
all_students = python_students | java_students
````
