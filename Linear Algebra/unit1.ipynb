## Lesson 1.1 — What is Linear Algebra?

Linear Algebra is a branch of mathematics that studies **vectors, matrices, linear equations, and linear transformations**.

It provides mathematical tools for representing and solving problems involving quantities, relationships, and transformations.

In this course, we will learn Linear Algebra step by step, beginning with the basic concepts and gradually moving to more advanced topics such as vector spaces, linear transformations, eigenvalues, eigenvectors, and diagonalization.

## 2. What is Linear Algebra?

Linear Algebra is a branch of mathematics concerned primarily with:

- Scalars
- Vectors
- Matrices
- Linear equations
- Vector spaces
- Linear transformations

Linear Algebra allows us to represent mathematical relationships in an organized form and solve problems involving multiple quantities.

For example, a system of equations such as:

x + y = 5

2x + 3y = 12

can be represented using matrices and vectors.

This makes it possible to use systematic methods such as:

- Gaussian elimination
- Gauss-Jordan elimination
- Matrix operations
- Determinants
- Matrix inverses

## 3. Why is it called "Linear" Algebra?

The word **linear** refers to relationships that can be represented using linear equations.

A simple linear equation can have the form:

ax + b = 0

where:

- a and b are constants
- x is a variable

For multiple variables, we can have equations such as:

2x + 3y = 10

or:

x + 2y + 3z = 15

These are linear equations because the variables are only raised to the first power.

Examples of linear expressions:

2x + 3y

x - 4y + 2z

5x + 7

Examples of expressions that are NOT linear:

x² + y

xy + 3

√x + y

1/x + y

The distinction between linear and nonlinear relationships will become important throughout this course.

## 5. Scalars

A scalar is a single numerical quantity.

Examples of scalars include:

5

-3

0

2.5

100

A scalar represents a single value rather than a collection of values.

We will use scalars frequently when working with vectors and matrices.

For example, if:

a = 5

then 5 is a scalar.

If we multiply a vector by a scalar, every component of the vector is multiplied by that scalar.

## 6. Vectors

A vector is an ordered collection of numbers.

For example:

v = [2, 5]

is a vector containing two components.

Another example is:

v = [3, 4, 7]

which contains three components.

Vectors can be written as row vectors or column vectors.

Row vector:

v = [2 5 7]

Column vector:

v =
[2]
[5]
[7]

The numbers inside a vector are called its components or entries.

Vectors are important because they can represent quantities such as:

- Position
- Direction
- Velocity
- Force
- Features in a dataset
- Numerical information

## 7. Matrices

A matrix is a rectangular arrangement of numbers organized into rows and columns.

For example:

A =
[1  2  3]
[4  5  6]

This matrix has:

- 2 rows
- 3 columns

Therefore, its dimension or order is:

2 × 3

A matrix can be used to organize numerical information and represent mathematical relationships.

Matrices are one of the most important objects in Linear Algebra.

Later in the course, we will learn:

- Matrix addition
- Matrix subtraction
- Scalar multiplication
- Matrix multiplication
- Transpose
- Determinants
- Matrix inverses
- Row operations
- Row echelon form (REF)
- Reduced row echelon form (RREF)

## 8. Linear Equations

A linear equation is an equation in which the variables have degree 1.

Examples include:

2x + 5 = 15

x + y = 10

2x + 3y = 12

x + 2y + 4z = 20

A system of linear equations consists of two or more linear equations considered together.

For example:

x + y = 5

2x + 3y = 12

The goal may be to find values of x and y that satisfy both equations simultaneously.

Linear Algebra provides systematic methods for solving such systems.

## 10. Connecting Scalars, Vectors, Matrices and Equations

We have now introduced four important ideas:

### Scalar
A single number.

Example:

5

### Vector
An ordered collection of numbers.

Example:

[2, 3]

### Matrix
A rectangular arrangement of numbers in rows and columns.

Example:

[1  2]
[3  4]

### Linear Equation
An equation involving variables with degree 1.

Example:

2x + 3y = 10

These concepts are closely connected.

A system of linear equations can be represented using matrices and vectors.

For example:

x + y = 5
2x + 3y = 12

can be written as:

A x = b

where:

A is the coefficient matrix,

x is the vector of unknown variables,

and b is the vector containing the constants.
